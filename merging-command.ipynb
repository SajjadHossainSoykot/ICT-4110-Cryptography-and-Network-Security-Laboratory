```python
import nbformat

files = [
    "Exp1_CeaserCipher.ipynb",
    "Exp2_PlayfairCipher.ipynb",
    "Exp3_HillCipher.ipynb",
    "Exp4_VigenereCipher.ipynb",
    "Exp5_RailFenceCipher.ipynb",
    "Exp6_RSA.ipynb",
    "Exp7_DiffeHellmanKeyExchange.ipynb"
]

merged = nbformat.v4.new_notebook()

for file in files:
    with open(file, "r", encoding="utf-8") as f:
        notebook = nbformat.read(f, as_version=4)
        merged.cells.extend(notebook.cells)

with open("All_Cryptography_Lab.ipynb", "w", encoding="utf-8") as f:
    nbformat.write(merged, f)

print("Notebooks merged successfully!")
```
